# 2. TOP 10

--->> Para empezar analizaré el df_antiguo que es en el que más artistas y géneros he conservado. 

```python
import matplotlib.pyplot as plt


# Cuento reproducciones por artista
artistas_top = df_antiguo['artista'].value_counts().head(10)  # Top 10 artistas

# Cuento reproducciones por género
generos_top = df_antiguo['género'].value_counts().head(10)  # Top 10 géneros

# Cuento reproducciones por canción 

canciones_top = df_antiguo['cancion'].value_counts().head(10)  # Top 10 canción

# Muestro los artistas, géneros y canciones más escuchadas
print("Artistas más escuchados:")
print(artistas_top)

print("\nGéneros más escuchados:")
print(generos_top)


print("\nCanciones más escuchadas:")
print(canciones_top)

# Configuro las variables en el gráfico los artistas más escuchados
plt.figure(figsize=(10, 5))
artistas_top.plot(kind='bar', color='violet')
plt.title('Top 10 Artistas Más Escuchados')
plt.xlabel('Artista')
plt.ylabel('Número de Reproducciones')
plt.xticks(rotation=45)
plt.show()

# Configuro las variables en el gráfico los géneros más escuchados
plt.figure(figsize=(10, 5))
generos_top.plot(kind='bar', color='darkmagenta')
plt.title('Top géneros más escuchados')
plt.xlabel('Artista')
plt.ylabel('Número de Reproducciones')
plt.xticks(rotation=45)
plt.show()

# Configuro las canciones más escuchadas
plt.figure(figsize=(10, 5))
canciones_top.plot(kind='bar', color='crimson')
plt.title('Top 10 Canciones Más Escuchadas')
plt.xlabel('Canción')
plt.ylabel('Número de Reproducciones')
plt.xticks(rotation=45)
plt.show()
```

# 3. ANÁLISIS DEL PATRÓN DE LA MÚSICA QUE ESCUCHO.

--->>> Resumen

* 1️. Top 10 artistas más escuchados (gráfico de barras).
* 2️. Día de la semana con más reproducciones (gráfico de barras).
* 3️. Evolución de la escucha a lo largo del tiempo (gráfico de líneas).
* 4️. Impacto de la temperatura en la escucha de música (gráfico de dispersión) -> para este he utilizado el último csv combinando la infrmación de Meteostat

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Asegurar que la columna fecha está en formato datetime en ambos DataFrames
df["fecha"] = pd.to_datetime(df["fecha"])
df_antiguo["fecha"] = pd.to_datetime(df_antiguo["fecha"])


# Distribución de escucha por día de la semana
plt.figure(figsize=(10,5))
df_antiguo["dia_semana"] = df_antiguo["fecha"].dt.day_name()

sns.countplot(x=df_antiguo["dia_semana"], order=["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"], palette="magma")
plt.xlabel("Día de la Semana")
plt.ylabel("Número de reproducciones")
plt.title("Escucha de música según el día de la semana")
plt.xticks(rotation=45)
plt.show()

# Evolución de la escucha a lo largo del tiempo
plt.figure(figsize=(14,6))
df_diario = df_antiguo.groupby("fecha").size()

sns.lineplot(x=df_diario.index, y=df_diario.values, color="blue")
plt.xlabel("Fecha")
plt.ylabel("Número de canciones escuchadas")
plt.title("Evolución de la escucha de música a lo largo del tiempo")
plt.show()
```

```python
--->> RELACIÓN ENTRE LA TEMPERATURA Y LA MÚSICA QUE ESCUCHO
```

```python
# Creo el gráfico
plt.figure(figsize=(12, 8))
plt.plot(df['fecha'], df['tavg_y'], label='Temperatura Promedio', color='blue', marker='o')
plt.plot(df['fecha'], df['tmax_y'], label='Temperatura Máxima', color='red', linestyle='--')
plt.plot(df['fecha'], df['tmin_y'], label='Temperatura Mínima', color='green', linestyle='--')
plt.scatter(df['fecha'], df['cancion'], label='Música Escuchada', color='purple', s=100, alpha=0.7)

# Personalizo del gráfico
plt.title('Relación entre la temperatura y la música escuchada', fontsize=16)
plt.xlabel('Fecha', fontsize=14)
plt.ylabel('Temperatura (°C) / Música Escuchada (minutos)', fontsize=14)
plt.xticks(fontsize=12, rotation=45)
plt.yticks(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()



# Muestro el gráfico
plt.show()
```

#### El gráfico muestra la evolución de la temperatura (máxima, mínima y promedio) a lo largo del tiempo junto con los patrones de escucha de música.
* Las temperaturas máxima, mínima y promedio están representadas con líneas de distintos colores (azul, rojo y verde).
* Las canciones escuchadas están marcadas con puntos morados, con los títulos de las canciones visibles en el eje vertical.
    * En este gráfico se puede ver que el día que he escuchado más música ha sido cuando la temperatura era más baja.
    * El día que presenta una temperatura más alta  he escuchado menos canciones
    * Algunas fechas tienen un aumento repentino en la escucha de música.
    * Es posible que ciertos eventos, como viajes, conciertos o estados de ánimo influidos por el clima, hayan influido en la cantidad de música reproducida.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Aseguro que la columna de fecha es de tipo datetime
df_antiguo["fecha"] = pd.to_datetime(df_antiguo["fecha"])

# Extraigo el mes de la fecha
df_antiguo["mes"] = df_antiguo["fecha"].dt.strftime("%Y-%m")

# Cuento el número de canciones escuchadas por mes
conteo_mensual = df_antiguo["mes"].value_counts().sort_index()

# Configuro el gráfico
plt.figure(figsize=(12, 6))
sns.barplot(x=conteo_mensual.index, y=conteo_mensual.values, palette="magma")

# Ajusto etiquetas
plt.xticks(rotation=45, ha="right")
plt.xlabel("Mes")
plt.ylabel("Número de canciones escuchadas")
plt.title("Distribución de música escuchada por mes")

plt.show()
```

### Análisis del patrón de escucha de música por mes: 

El gráfico muestra la distribución de la música escuchada a lo largo del tiempo, agrupada por meses. A partir de esta visualización, se pueden extraer varios patrones interesantes:

* Pico significativo en abril de 2023
	•	Se observa un aumento notable en el número de canciones escuchadas en abril de 2023, alcanzando su punto más alto en todo el período analizado.
	•	Esto podría estar relacionado con factores como vacaciones, justo fue ese año cuando las tuve en esa fecha y posiblemente escuchara más música en el viaje

* Variabilidad en el consumo musical
	•	Después del pico de abril de 2023, hay una fluctuación en la cantidad de música escuchada, con algunos meses con niveles relativamente bajos y otros con repuntes.
	•	Hay una recuperación en abril y mayo de 2024, aunque no tan alta como en 2023.

* Tendencia a la baja en los últimos meses
	•	A partir de mediados de 2024, parece haber una disminución progresiva en la cantidad de música escuchada.
	•	Esto podría indicar un cambio en hábitos de consumo o menor disponibilidad de tiempo para escuchar música, seguramente que por el máster que me está quitando más tiempo libre.
